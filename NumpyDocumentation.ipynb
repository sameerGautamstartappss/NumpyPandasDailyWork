One magor difference is that slice indexing of a `list` copies the elements into a **new list**, but slicing an `array` returns a **`view`**: an object that refers to the data in the original array.

Python List: When you slice a list, it creates a **new, independent copy** of the data.
means:
# The new list(created by slicing) **contains the same elements** as the original.
# But it is stored **in a different place in the memory**, and
# Changing the sliced list does NOT affect the original list.

In NumPy, a dimension of an array is sometimes referred to as an "axis".


```python
# Priniting all elements

for i in range(row):
    for j in range(col):
        print(arr[i, j])
```

# Unlike a list of lists where you'd write list[i][j], numpy allows arr[i, j]


# Axis = 0 => row wise => refers to vertical direction => operation will be performed row wise (no operation column wise) => (for example, add lists)

# Axis = 1 => operation within seperate lists. Axis refers to column wise operation. Horrizontal direction(columns). Axis 1 refers to horrizontal direction.

# In case of three dimensions,

# axis = 0 is depth (stacked layers)
# axis = 1 is rows 
# axis = 2 is columns

```python
# In NumPy, what does axis = 0 represent when working with a 2D arrays ?

arr = np.array([[2,4], [6,8]])
```

```python
# How does axes differ when working with a 3D array compared to a 2D array?

# Think of axes as dimensions, and each additional dimensions adds a level of nesting.

# 2D Array (shape: (rows, columns))
```

```python
# Example
row = int(input("Enter the number of lists : "))
col = int(input("Enter the number of elements in each list : "))
arr = np.array((lambda row, col : list(list(int(input(f"Enter the {j+1} row and {i+1} column element: ")) for i in range(col)) for j in range(row)))(row, col))
```

```python
arr
```

```python
type(arr)
```

```python
np.sum(arr, axis=0)
```

```python
np.sum(arr, axis=1)
```

```python
# 3D Array
# Think of this is as a stack of 2D matrices
```

```python
row = int(input("Enter the number of elements in the row : "))
col = int(input("Enter the number of elements in the column : "))
depth = int(input("Enter the number of flights : "))
three_dimensional_array = np.array((lambda row, col, depth : list(list(list(int(input(f"Enter the {k+1} flight, {j+1} row and {i+1} column element : ")) for i in range(col)) for j in range(row)) for k in range(depth)))(row, col, depth))
```
